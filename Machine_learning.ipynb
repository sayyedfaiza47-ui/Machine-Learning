```python
hp_mean = car_data2['Engine_HP'].mean()
hp_mean
```

```python
car_data2['Engine_HP'] = car_data2['Engine_HP'].fillna(hp_mean)
```

```python
car_data2.isnull().sum()
```

```python
No_of_Cylinders_mean = car_data2['No_of_Cylinders'].mean()
car_data2['No_of_Cylinders'] = car_data2['No_of_Cylinders'].fillna(No_of_Cylinders_mean)
```

```python
car_data2.isnull().sum()
```

```python
No_of_Doors_mean = car_data2['No_of_Doors'].mean()
car_data2['No_of_Doors'] = car_data2['No_of_Doors'].fillna(No_of_Doors_mean)
```

```python
car_data2.isnull().sum()
```

```python
Market_Category_mode = car_data2['Market_Category'].mode()[0]
car_data2['Market_Category'] = car_data2['Market_Category'].fillna(Market_Category_mode)
```
